## Visualizando Distribuições

Normalmente é usado o histograma para visualizar a frequencia dos dados
    - permite ver se o dados seguem uma distribuição normal (Gaussiana)
    - identificar associametrias (skewness)
    - revela a curtose (concentração dos dados)

## Identificando dados faltantes
Missing values

O problema dos nulls
Dados ausentes podem introduzir viés ou causar erros em algoritmos que não suportam entradas vazias

detecção
```python

df.isnull().sum()

```

Retorna a contagem de valores nulos por coluna.

## Caçador de outliers
Boxplot

O boxplot resume a distribuição através dos quartis:
- Caixa: representa 50% dos dados centrais
- Linha central: mediana
- Pontos externos: outlier potenciais
